```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Generate dataset with 10 samples
data = {
    'education_level': np.random.choice(['Primary', 'Secondary', 'Tertiary', 'No Formal Education'], 10),
    'age': np.random.randint(18, 75, 10),
    'gender': np.random.choice(['Male', 'Female'], 10)
}

# Create DataFrame
df = pd.DataFrame(data)
print(df)
```
